```python
import os
import re
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from sklearn.linear_model import Ridge, Lasso, ElasticNet, BayesianRidge
from sklearn.cross_decomposition import PLSRegression
from sklearn.svm import SVR
from sklearn.ensemble import HistGradientBoostingRegressor, StackingRegressor

import joblib

# ============================================================
# 0) CONFIG: Inputs/Outputs (IGNORE output_2 entirely)
# ============================================================
input_columns = [
    '9711900_a', '9710700_a', '9710702_a', '9710190_a',
    '9710703_a', '9712055_a', '9710293_a', '9710278_a',
    '9710709_a', '9710475_a', '9710275_a',
    'Cal_a', 'Talc_a', '9801084_a', '9710146_a',
    'PEA0330648_a', 'PP070G2M_a', 'BE961MO_a',
    '9710156_a', '9710169_a',
    'Beyaz_a', 'Gri_a',
]

output_columns = [
    'Density_a',
    'Tensile Strength (Mpa)_a',
    'Elongation at Yield (%)_a',
    'Break Strength (Mpa)_a',
    'Izod İmpact_a',
    'Flexural Strength (Mpa)_a'
]

# IMPORTANT: we do NOT use output_2 at all
output_2 = ['Elastic Modulus (Mpa)_a', 'Elongation at Break (%)_a', 'Flexural Modulus (Mpa)_a']

# ============================================================
# 1) Helpers
# ============================================================
def safe_name(s: str) -> str:
    s = str(s)
    s = s.strip()
    s = re.sub(r"[^\w\-\.\(\)%]+", "_", s)
    return s

def assert_columns_exist(df, cols, name="columns"):
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise KeyError(f"Missing {name} in df: {missing}")

# ============================================================
# 2) Data: assume df is ALREADY prepared (no scaler/imputer here)
# ============================================================
df = df.copy()
assert_columns_exist(df, input_columns, "input_columns")
assert_columns_exist(df, output_columns, "output_columns")

X = df[input_columns].copy()
Y = df[output_columns].copy()

# Ensure numeric (does not impute/scale; just coerces)
X = X.apply(pd.to_numeric, errors="coerce")
Y = Y.apply(pd.to_numeric, errors="coerce")

# If you truly have NO missing values, the next lines should be 0
na_x = int(X.isna().sum().sum())
na_y = int(Y.isna().sum().sum())
print(f"NaNs in X: {na_x} | NaNs in Y: {na_y}")
if na_x > 0 or na_y > 0:
    print("WARNING: You said data is already prepared. But NaNs exist. "
          "Either fix upstream or drop rows here (example below).")
    # Option (uncomment) to drop rows with any NaNs:
    # mask = ~(X.isna().any(axis=1) | Y.isna().any(axis=1))
    # X, Y = X.loc[mask], Y.loc[mask]

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

# ============================================================
# 3) Model builders (NO scaler, NO imputer)
# ============================================================
def make_ridge():
    return Ridge(random_state=42)

def make_lasso():
    return Lasso(max_iter=20000, random_state=42)

def make_elasticnet():
    return ElasticNet(max_iter=20000, random_state=42)

def make_poly2_elasticnet():
    return Pipeline(steps=[
        ("poly", PolynomialFeatures(degree=2, include_bias=False)),
        ("model", ElasticNet(max_iter=20000, random_state=42)),
    ])

def make_bayesian_ridge():
    return BayesianRidge()

def make_pls():
    return PLSRegression()

def make_svr_rbf():
    return SVR(kernel="rbf")

def make_histgb():
    return HistGradientBoostingRegressor(random_state=42)

def make_stacking():
    # Base estimators are pipelines / estimators with NO scaler/imputer (assumes X already prepared)
    base_estimators = [
        ("svr", make_svr_rbf()),
        ("bayes", make_bayesian_ridge()),
        ("pls", make_pls()),
        ("poly_en", make_poly2_elasticnet()),
    ]
    final_estimator = Ridge(alpha=1.0, random_state=42)

    return StackingRegressor(
        estimators=base_estimators,
        final_estimator=final_estimator,
        passthrough=False,
        n_jobs=-1
    )

# ============================================================
# 4) Hyperparameter search spaces
#    - GridSearch for linear models / poly
#    - RandomizedSearch for SVR / HistGB
# ============================================================
SEARCHES = {
    "Ridge": ("grid", make_ridge(), {
        "alpha": np.logspace(-4, 4, 25)
    }),

    "Lasso": ("grid", make_lasso(), {
        "alpha": np.logspace(-6, 1, 25)
    }),

    "ElasticNet": ("grid", make_elasticnet(), {
        "alpha": np.logspace(-6, 1, 20),
        "l1_ratio": [0.1, 0.3, 0.5, 0.7, 0.9]
    }),

    "Poly2_ElasticNet": ("grid", make_poly2_elasticnet(), {
        "model__alpha": np.logspace(-6, 1, 20),
        "model__l1_ratio": [0.1, 0.3, 0.5, 0.7, 0.9]
    }),

    "BayesianRidge": ("grid", make_bayesian_ridge(), {
        # BayesianRidge has fewer impactful knobs; keep it light
        "alpha_1": [1e-6, 1e-5, 1e-4],
        "alpha_2": [1e-6, 1e-5, 1e-4],
        "lambda_1": [1e-6, 1e-5, 1e-4],
        "lambda_2": [1e-6, 1e-5, 1e-4],
    }),

    "PLSRegression": ("grid", make_pls(), {
        "n_components": [2, 3, 4, 5, 6, 8, 10]
    }),

    "SVR_RBF": ("random", make_svr_rbf(), {
        "C": np.logspace(0, 4, 100),
        "gamma": np.logspace(-6, 0, 100),
        "epsilon": np.logspace(-3, 0, 50)
    }),

    "HistGB": ("random", make_histgb(), {
        "learning_rate": np.logspace(-3, -0.3, 60),
        "max_depth": [2, 3, 4, 5, 6, None],
        "max_leaf_nodes": [15, 31, 63, 127],
        "min_samples_leaf": [5, 10, 20, 30],
        "l2_regularization": np.logspace(-3, 2, 50)
    }),

    "Stacking": ("grid", make_stacking(), {
        # keep light; stacking is heavy
        "final_estimator__alpha": np.logspace(-4, 3, 15)
    }),
}

# ============================================================
# 5) Train, pick BEST per target, save models
# ============================================================
os.makedirs("saved_models", exist_ok=True)

best_models = {}
rows = []

for target in output_columns:
    ytr = Y_train[target]
    yte = Y_test[target]

    print("\n" + "="*70)
    print(f"TARGET: {target}")
    print("="*70)

    best_r2 = -np.inf
    best_entry = None  # (name, estimator, best_params, metrics)

    for model_name, (kind, est, params) in SEARCHES.items():
        if kind == "grid":
            search = GridSearchCV(
                est, params, cv=5, scoring="r2", n_jobs=-1, verbose=0
            )
        else:
            search = RandomizedSearchCV(
                est, params, n_iter=40, cv=5, scoring="r2",
                n_jobs=-1, random_state=42, verbose=0
            )

        search.fit(X_train, ytr)
        best_est = search.best_estimator_

        pred = best_est.predict(X_test)
        r2 = r2_score(yte, pred)
        mae = mean_absolute_error(yte, pred)
        rmse = mean_squared_error(yte, pred)

        rows.append([target, model_name, r2, mae, rmse, search.best_params_])

        print(f"{model_name:<16} | R2={r2:>8.4f} | MAE={mae:>10.4f} | RMSE={rmse:>10.4f}")

        if r2 > best_r2:
            best_r2 = r2
            best_entry = (model_name, best_est, search.best_params_, (r2, mae, rmse))

    # Save best model for this target
    best_name, best_estimator, best_params, (r2, mae, rmse) = best_entry
    best_models[target] = {"model": best_estimator, "name": best_name, "params": best_params, "metrics": (r2, mae, rmse)}

    model_path = os.path.join("saved_models", f"best_{safe_name(target)}.joblib")
    joblib.dump(best_estimator, model_path)

    print(f"\n>>> BEST for {target}: {best_name} | R2={r2:.4f} | saved to {model_path}")

# ============================================================
# 6) Summary tables
# ============================================================
results_df = pd.DataFrame(rows, columns=["Target", "Model", "R2", "MAE", "RMSE", "BestParams"])
results_df = results_df.sort_values(["Target", "R2"], ascending=[True, False])

print("\n" + "#"*70)
print("ALL MODEL RESULTS (sorted by Target then R2)")
print("#"*70)
print(results_df.to_string(index=False))

best_summary = []
for t, info in best_models.items():
    r2, mae, rmse = info["metrics"]
    best_summary.append([t, info["name"], r2, mae, rmse, info["params"]])

best_df = pd.DataFrame(best_summary, columns=["Target", "BestModel", "R2", "MAE", "RMSE", "BestParams"])
best_df = best_df.sort_values("R2", ascending=False)

print("\n" + "#"*70)
print("BEST MODEL PER TARGET (saved under ./saved_models)")
print("#"*70)
print(best_df.to_string(index=False))

# Optional: save summary tables
best_df.to_csv("best_models_summary.csv", index=False)
results_df.to_csv("all_models_results.csv", index=False)
print("\nSaved: best_models_summary.csv, all_models_results.csv")
```

```python
import os
import re
import numpy as np
import pandas as pd

from sklearn.model_selection import KFold, GridSearchCV, RandomizedSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

from sklearn.linear_model import Ridge, Lasso, ElasticNet, BayesianRidge
from sklearn.cross_decomposition import PLSRegression
from sklearn.svm import SVR
from sklearn.ensemble import HistGradientBoostingRegressor, StackingRegressor

import joblib

# ============================================================
# STAGE-2: Predict output_2 using df_output_2
# - Inputs: all columns except output_2
# - Outputs: Elastic Modulus, Elongation at Break, Flexural Modulus
# - NO scaler, NO imputer (data already prepared)
# - Use 5-fold CV for model selection
# - Save best model per target
# ============================================================

output_2 = [
    'Elastic Modulus (Mpa)_a',
    'Elongation at Break (%)_a',
    'Flexural Modulus (Mpa)_a',
]

df_stage2 = df_output_2.copy()

# Ensure numeric (no imputation/scaling; just coercion)
df_stage2 = df_stage2.apply(pd.to_numeric, errors="coerce")

# Split X2 / Y2
X2 = df_stage2.drop(columns=output_2)
Y2 = df_stage2[output_2]

# Optional: if any NaNs exist, drop rows (recommended since no imputer)
mask = ~(X2.isna().any(axis=1) | Y2.isna().any(axis=1))
X2 = X2.loc[mask].copy()
Y2 = Y2.loc[mask].copy()

print("Stage-2 shapes:", X2.shape, Y2.shape)

# Train/test split only for final holdout evaluation (CV still used for selection)
from sklearn.model_selection import train_test_split
X2_train, X2_test, Y2_train, Y2_test = train_test_split(
    X2, Y2, test_size=0.2, random_state=42
)

# 5-fold CV setup
cv5 = KFold(n_splits=5, shuffle=True, random_state=42)

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def safe_name(s: str) -> str:
    s = str(s).strip()
    s = re.sub(r"[^\w\-\.\(\)%]+", "_", s)
    return s

# ------------------------------------------------------------
# Model builders (NO scaler, NO imputer)
# ------------------------------------------------------------
def make_ridge():
    return Ridge(random_state=42)

def make_lasso():
    return Lasso(max_iter=30000, random_state=42)

def make_elasticnet():
    return ElasticNet(max_iter=30000, random_state=42)

def make_poly2_elasticnet():
    return Pipeline(steps=[
        ("poly", PolynomialFeatures(degree=2, include_bias=False)),
        ("model", ElasticNet(max_iter=30000, random_state=42)),
    ])

def make_bayesian_ridge():
    return BayesianRidge()

def make_pls():
    return PLSRegression()

def make_svr_rbf():
    return SVR(kernel="rbf")

def make_histgb():
    return HistGradientBoostingRegressor(random_state=42)

def make_stacking():
    # Base estimators (no preprocessing inside)
    base_estimators = [
        ("svr", make_svr_rbf()),
        ("bayes", make_bayesian_ridge()),
        ("pls", make_pls()),
        ("poly_en", make_poly2_elasticnet()),
    ]
    final_estimator = Ridge(alpha=1.0, random_state=42)
    return StackingRegressor(
        estimators=base_estimators,
        final_estimator=final_estimator,
        passthrough=False,
        n_jobs=-1
    )

# ------------------------------------------------------------
# Search spaces
# ------------------------------------------------------------
SEARCHES = {
    "Ridge": ("grid", make_ridge(), {
        "alpha": np.logspace(-4, 4, 25)
    }),
    "Lasso": ("grid", make_lasso(), {
        "alpha": np.logspace(-6, 1, 25)
    }),
    "ElasticNet": ("grid", make_elasticnet(), {
        "alpha": np.logspace(-6, 1, 20),
        "l1_ratio": [0.1, 0.3, 0.5, 0.7, 0.9]
    }),
    "Poly2_ElasticNet": ("grid", make_poly2_elasticnet(), {
        "model__alpha": np.logspace(-6, 1, 20),
        "model__l1_ratio": [0.1, 0.3, 0.5, 0.7, 0.9]
    }),
    "BayesianRidge": ("grid", make_bayesian_ridge(), {
        "alpha_1": [1e-6, 1e-5, 1e-4],
        "alpha_2": [1e-6, 1e-5, 1e-4],
        "lambda_1": [1e-6, 1e-5, 1e-4],
        "lambda_2": [1e-6, 1e-5, 1e-4],
    }),
    "PLSRegression": ("grid", make_pls(), {
        "n_components": [2, 3, 4, 5, 6, 8, 10]
    }),
    "SVR_RBF": ("random", make_svr_rbf(), {
        "C": np.logspace(0, 4, 150),
        "gamma": np.logspace(-6, 0, 150),
        "epsilon": np.logspace(-3, 0, 60)
    }),
    "HistGB": ("random", make_histgb(), {
        "learning_rate": np.logspace(-3, -0.3, 80),
        "max_depth": [2, 3, 4, 5, 6, None],
        "max_leaf_nodes": [15, 31, 63, 127],
        "min_samples_leaf": [5, 10, 20, 30],
        "l2_regularization": np.logspace(-3, 2, 80)
    }),
    "Stacking": ("grid", make_stacking(), {
        "final_estimator__alpha": np.logspace(-4, 3, 15)
    }),
}

# ------------------------------------------------------------
# Train per target with 5-fold CV selection, save best models
# ------------------------------------------------------------
os.makedirs("saved_models_stage2", exist_ok=True)

best_models = {}
rows = []

for target in output_2:
    ytr = Y2_train[target]
    yte = Y2_test[target]

    print("\n" + "="*80)
    print(f"STAGE-2 TARGET: {target}")
    print("="*80)

    best_r2 = -np.inf
    best_entry = None  # (model_name, best_estimator, best_params, metrics)

    for model_name, (kind, est, params) in SEARCHES.items():
        if kind == "grid":
            search = GridSearchCV(
                est, params, cv=cv5, scoring="r2", n_jobs=-1, verbose=0
            )
        else:
            search = RandomizedSearchCV(
                est, params, n_iter=50, cv=cv5, scoring="r2",
                n_jobs=-1, random_state=42, verbose=0
            )

        search.fit(X2_train, ytr)
        best_est = search.best_estimator_

        # Holdout evaluation (optional but useful)
        pred = best_est.predict(X2_test)
        r2 = r2_score(yte, pred)
        mae = mean_absolute_error(yte, pred)
        rmse = mean_squared_error(yte, pred)

        rows.append([target, model_name, r2, mae, rmse, search.best_params_])

        print(f"{model_name:<16} | Holdout R2={r2:>8.4f} | MAE={mae:>10.4f} | RMSE={rmse:>10.4f}")

        if r2 > best_r2:
            best_r2 = r2
            best_entry = (model_name, best_est, search.best_params_, (r2, mae, rmse))

    best_name, best_estimator, best_params, (r2, mae, rmse) = best_entry
    best_models[target] = {"model": best_estimator, "name": best_name, "params": best_params, "metrics": (r2, mae, rmse)}

    # Save best model
    model_path = os.path.join("saved_models_stage2", f"best_{safe_name(target)}.joblib")
    joblib.dump(best_estimator, model_path)

    print(f"\n>>> BEST for {target}: {best_name} | Holdout R2={r2:.4f} | saved to {model_path}")

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------
results_df = pd.DataFrame(rows, columns=["Target", "Model", "Holdout_R2", "MAE", "RMSE", "BestParams"])
results_df = results_df.sort_values(["Target", "Holdout_R2"], ascending=[True, False])

print("\n" + "#"*80)
print("STAGE-2 ALL RESULTS (sorted by Target then Holdout_R2)")
print("#"*80)
print(results_df.to_string(index=False))

best_summary = []
for t, info in best_models.items():
    r2, mae, rmse = info["metrics"]
    best_summary.append([t, info["name"], r2, mae, rmse, info["params"]])

best_df = pd.DataFrame(best_summary, columns=["Target", "BestModel", "Holdout_R2", "MAE", "RMSE", "BestParams"])
best_df = best_df.sort_values("Holdout_R2", ascending=False)

print("\n" + "#"*80)
print("STAGE-2 BEST MODEL PER TARGET (saved under ./saved_models_stage2)")
print("#"*80)
print(best_df.to_string(index=False))

best_df.to_csv("best_models_stage2_summary.csv", index=False)
results_df.to_csv("all_models_stage2_results.csv", index=False)
print("\nSaved: best_models_stage2_summary.csv, all_models_stage2_results.csv")
```

```python
import numpy as np
import pandas as pd
import joblib
import re

# ============================================================
# PURPOSE
# - Your stage-2 model expects engineered columns like:
#   Break Strength (Mpa)_a__pred_over_density
#   Density_a__pred_x_Break Strength (Mpa)_a__pred
#   ...
# - This code:
#   1) asks for the 22 base inputs
#   2) predicts the 6 "__pred" values (stage-1 models)
#   3) creates ALL engineered features needed for stage-2
#   4) aligns feature columns to exactly what stage-2 models expect
#   5) predicts the last 3 outputs
# ============================================================

# -----------------------------
# Load models (update paths if needed)
# -----------------------------
MODEL_PATHS = {
    # ---- Stage-1 outputs (predicted first) ----
    "Density_a__pred": "/content/best_Density_a.joblib",
    "Tensile Strength (Mpa)_a__pred": "/content/best_Tensile_Strength_(Mpa)_a.joblib",
    "Elongation at Yield (%)_a__pred": "/content/best_Elongation_at_Yield_(%)_a.joblib",
    "Break Strength (Mpa)_a__pred": "/content/best_Break_Strength_(Mpa)_a.joblib",
    "Izod İmpact_a__pred": "/content/best_Izod_İmpact_a.joblib",
    "Flexural Strength (Mpa)_a__pred": "/content/best_Flexural_Strength_(Mpa)_a.joblib",

    # ---- Stage-2 outputs (predicted second) ----
    "Elastic Modulus (Mpa)_a": "/content/best_Elastic_Modulus_(Mpa)_a.joblib",
    "Elongation at Break (%)_a": "/content/best_Elongation_at_Break_(%)_a.joblib",
    "Flexural Modulus (Mpa)_a": "/content/best_Flexural_Modulus_(Mpa)_a.joblib",
}

models = {k: joblib.load(v) for k, v in MODEL_PATHS.items()}

# -----------------------------
# Base inputs (22)
# -----------------------------
input_columns = [
    '9711900_a', '9710700_a', '9710702_a', '9710190_a',
    '9710703_a', '9712055_a', '9710293_a', '9710278_a',
    '9710709_a', '9710475_a', '9710275_a',
    'Cal_a', 'Talc_a', '9801084_a', '9710146_a',
    'PEA0330648_a', 'PP070G2M_a', 'BE961MO_a',
    '9710156_a', '9710169_a',
    'Beyaz_a', 'Gri_a',
]

stage1_pred_cols = [
    "Density_a__pred",
    "Tensile Strength (Mpa)_a__pred",
    "Elongation at Yield (%)_a__pred",
    "Break Strength (Mpa)_a__pred",
    "Izod İmpact_a__pred",
    "Flexural Strength (Mpa)_a__pred",
]

stage2_targets = [
    "Elastic Modulus (Mpa)_a",
    "Elongation at Break (%)_a",
    "Flexural Modulus (Mpa)_a",
]

# -----------------------------
# Ask user inputs
# -----------------------------
def ask_inputs() -> pd.DataFrame:
    print("\nEnter the 22 input values (numbers). For Beyaz_a/Gri_a use 0 or 1.\n")
    row = {}
    for col in input_columns:
        while True:
            val = input(f"{col}: ").strip()
            try:
                row[col] = float(val)
                break
            except ValueError:
                print("  ❌ Please enter a numeric value.")
    return pd.DataFrame([row], columns=input_columns)

# -----------------------------
# Feature engineering for Stage-2
#   (creates the missing *_x_* and *_over_density etc.)
# -----------------------------
def build_stage2_features(df_base: pd.DataFrame) -> pd.DataFrame:
    """
    df_base: one-row DF containing input_columns + stage1_pred_cols
    returns: engineered DF containing base + engineered features
    """
    eps = 1e-12
    df = df_base.copy().apply(pd.to_numeric, errors="coerce")

    # composition features from original inputs (optional but harmless)
    comp_cols = [c for c in input_columns]  # treat all 22 as composition-like
    df["comp_sum"] = df[comp_cols].sum(axis=1)
    df["comp_nonzero_count"] = (df[comp_cols].fillna(0).abs() > 0).sum(axis=1)
    df["comp_mean_nonzero"] = df["comp_sum"] / (df["comp_nonzero_count"] + eps)
    df["comp_std"] = df[comp_cols].std(axis=1)
    df["comp_cv"] = df["comp_std"] / (df[comp_cols].mean(axis=1) + eps)
    df["comp_max"] = df[comp_cols].max(axis=1)
    df["comp_max_share"] = df["comp_max"] / (df["comp_sum"] + eps)

    # filler features
    df["filler_sum"] = df["Cal_a"].fillna(0) + df["Talc_a"].fillna(0)
    df["talc_cal_ratio"] = df["Talc_a"] / (df["Cal_a"] + eps)
    df["filler_share"] = df["filler_sum"] / (df["comp_sum"] + eps)

    # color flags
    df["is_beyaz"] = (df["Beyaz_a"] > 0).astype(int)
    df["is_gri"] = (df["Gri_a"] > 0).astype(int)
    df["color_conflict"] = ((df["is_beyaz"] == 1) & (df["is_gri"] == 1)).astype(int)

    # log1p of fillers and totals
    df["log1p_Cal_a"] = np.log1p(df["Cal_a"].fillna(0).clip(lower=0))
    df["log1p_Talc_a"] = np.log1p(df["Talc_a"].fillna(0).clip(lower=0))
    df["log1p_filler_sum"] = np.log1p(df["filler_sum"].fillna(0).clip(lower=0))
    df["log1p_comp_sum"] = np.log1p(df["comp_sum"].fillna(0).clip(lower=0))

    # transforms for each stage-1 pred
    for c in stage1_pred_cols:
        df[f"log_{c}"] = np.log1p(df[c].clip(lower=0))
        df[f"sqr_{c}"] = df[c] ** 2

    # pairwise interactions among stage-1 preds (this creates the missing *_x_*)
    for i in range(len(stage1_pred_cols)):
        for j in range(i + 1, len(stage1_pred_cols)):
            c1, c2 = stage1_pred_cols[i], stage1_pred_cols[j]
            df[f"{c1}_x_{c2}"] = df[c1] * df[c2]

    # ratios over density (this creates *_over_density)
    for c in stage1_pred_cols:
        if c != "Density_a__pred":
            df[f"{c}_over_density"] = df[c] / (df["Density_a__pred"] + eps)

    return df

# -----------------------------
# Align features to what the stage-2 model expects
# (prevents "missing feature" errors)
# -----------------------------
def align_to_model_features(model, X: pd.DataFrame) -> pd.DataFrame:
    feat = getattr(model, "feature_names_in_", None)
    if feat is None:
        # fallback: model doesn't store names; assume current X order is fine
        return X
    feat = list(feat)
    X_aligned = X.copy()
    # add any missing columns as 0
    missing = [c for c in feat if c not in X_aligned.columns]
    for c in missing:
        X_aligned[c] = 0.0
    # drop extra columns and order exactly
    X_aligned = X_aligned[feat]
    return X_aligned

# -----------------------------
# Main prediction: 9 outputs
# -----------------------------
def predict_all_9():
    # 1) read base inputs
    X1 = ask_inputs()

    # 2) stage-1 predictions
    X1_num = X1.apply(pd.to_numeric, errors="coerce")
    for col in stage1_pred_cols:
        X1[col] = float(models[col].predict(X1_num)[0])

    # 3) stage-2 engineered features
    X2 = build_stage2_features(X1)

    # 4) predict final 3 with proper feature alignment
    final_preds = {}
    for tgt in stage2_targets:
        m = models[tgt]
        X2_aligned = align_to_model_features(m, X2)
        final_preds[tgt] = float(m.predict(X2_aligned)[0])

    # 5) collect all 9 predictions
    out = {col: float(X1[col].iloc[0]) for col in stage1_pred_cols}
    out.update(final_preds)

    out_df = pd.DataFrame([out])
    print("\n==================== PREDICTIONS ====================")
    print(out_df.T.rename(columns={0: "pred"}))
    return out_df

# Usage:
# preds_df = predict_all_9()
```
